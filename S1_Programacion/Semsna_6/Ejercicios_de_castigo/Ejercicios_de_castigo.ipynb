#### Ejercicio 13: Suma de una lista con función que regresa valor (sin `sum()`)

#### Ejercicio 14: Máximo de una lista con función que regresa valor (sin `max()`)
**Entrada:** `[7, 3, 9, 2]` → **Salida:** `El elemento máximo es: 9

````python
lista_num=[7,3,9,2]

def mayor(numero):
    numero_m=0
    for i in range (len(numero)):
        if numero[i]>numero_m:
            numero_m=numero[i]
    return numero_m
numero_m=mayor(lista_num)
print(f"el numero mayor de la lista es {numero_m}")
        
````

#### Ejercicio 15: Matriz y suma de la diagonal principal
**Entrada:** `[[1,2,3],[4,5,6],[7,8,9]]` → **Salida:** `La suma de la diagonal
principal es: 15`

````python
matriz=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
suma=0

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        if i == j: 
            suma += matriz[i][j]

print(f"La suma de la diagonal principal es: {suma}")
````

#### Ejercicio 20: Diccionario: buscar teléfono de un contacto
**Entrada:** `Beto` → **Salida:** `El teléfono de Beto es: 5555678`

````python
directorio = {
    "beto": "5555678"
}

def buscar_telefono(nombre):
    nombre_minúscula = nombre.lower()
    return directorio.get(nombre_minúscula, "no lo tengo pibe")

contacto = input("¿Qué contacto buscas?: ")
telefono = buscar_telefono(contacto)

if telefono != "no lo tengo pibe":
    print(f"El teléfono de {contacto} es: {telefono}")
else:
    print(telefono)

````

#### Ejercicio 21: Diccionario de calificaciones y promedios

````python
calificaciones = {
    "Ana": [8, 9, 10], 
    "Beto": [7, 7, 8]
}

for nombre, notas in calificaciones.items():
    promedio = sum(notas) / len(notas)
    
    print(f"{nombre}: {promedio:.2f}")
````
