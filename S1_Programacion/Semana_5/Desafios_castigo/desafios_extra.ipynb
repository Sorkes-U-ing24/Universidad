### Desafío Extra 2: Contador de votos con diccionario
Crea un diccionario `votos = {"Rojo": 0, "Azul": 0, "Verde": 0}`. Pide al usuario
el color de su voto cinco veces, suma uno al valor correspondiente en cada ocasión
y, al final, imprime con `items()` el color ganador junto con su total.

```python
candidatos={
    "A": 0,
    "B": 0,
    "C": 0,
    }

cantidad=int(input("cuantas personas votaran"))
conteo=1
while conteo<=cantidad:
     voto=input(f"ingresa el voto{conteo} (A, B, o C): ")
     if voto.upper() in candidatos:
         print(f"voto vlido para {voto.upper()}")
         candidatos[voto.upper()]+=1
         conteo+=1
     else:
         print("el voto no es valido")

for k, v in candidatos.items():
    print(f"\nOpcion {k}, votos: {v}")
         
```

### Ejercicio 1: Tercer elemento de una tupla
**Enunciado:** Dada la tupla `colores = ("rojo", "verde", "azul", "amarillo",
"naranja")`, imprime el tercer elemento (índice 2) y el primer elemento (índice 0).
**Entrada:** `colores = ("rojo", "verde", "azul", "amarillo", "naranja")`
**Salida:** `El tercer color es: azul` y `El primer color es: rojo`

```python
colores=("rojo", "verde", "azul", "amarillo", "naranja")

for i in range(len(colores)):
     if i==0:
          rojo=colores[i]
     if i==2:
         azul=colores[i]
print(f"primer color es {rojo}\ntercer color es {azul}")
```

### Desafío Extra 5: Minimenú modular integrador
Construye un menú con un ciclo `while` y opciones numéricas que permita ejecutar de
forma modular: (1) mostrar la tupla de números ordenada, (2) buscar un teléfono en
el diccionario `contactos`, (3) dividir dos números con manejo de
`ZeroDivisionError` y (4) analizar un mensaje de texto. La opción (5) debe
finalizar el programa.

```python
#opcion 1
tupla_numeros=(1, 63, 54, 95, 21, 74, 62, 35)

#opcion 2
diccionario_agenda = {
    "ANA": "555-0101", 
    "LUIS": "555-0102", 
    "MIA": "555-0103"
}

opcion=None

while opcion !=5:
    print("\n-----MENU-----")
    print("1. mostrar tupla ordenada")
    print("2. buscar telefono en diccionario")
    print("3. dividir dos numeros con excepciones")
    print("4. analizar un mensaje de texto")
    print("5. finalizar programa")
    
    try:
         opcion=int(input("Escoge una opcion: "))
    except:
        print("error, ecoge una opcion")
        continue
    if opcion==1:
        tupla_ordenada = tuple(sorted(tupla_numeros))
        print(f"Tupla original: {tupla_numeros}")
        print(f"Tupla ordenada: {tupla_ordenada}")
        
    elif opcion==2:
        nombres_validos=["ANA", "lUIS", "MIA"]
        while True:
             busqueda=input("el numero de quien quieres buscar?:\nAna\nLuis\nMia").upper
             if busqueda in nombres_validos:
                 break
             print("Error, pon un nombre de la lista")
             
        def contacto_busqueda(contactos):
            numero_contacto= diccionario_agenda.get(contactos, "numero no encontrado")
            return f"contacto: {contactos} | numero: {numero_contacto}"
        
        contacto_encontrado=contacto_busqueda(busqueda)
        print(f" el  contacto de {busqueda} es {contacto_encontrado} ")
        
    elif opcion == 3:
        while True:
            try: 
                num1 = int(input("Ingresa el primer número (dividendo): "))
                num2 = int(input("Ingresa el segundo número (divisor): "))
                
                # La división y su captura de ZeroDivisionError deben ir aquí
                resultado_de_division = num1 / num2
                print(f"Resultado de la división: {resultado_de_division}")
                break
            except ValueError:
                print("Error: Debes ingresar números enteros válidos.")
            except ZeroDivisionError:
                print("Error: No se puede dividir entre cero. Intenta de nuevo.")
                    
                    
    elif opcion == 4:
        frase = input("Ingresa una frase: ")
        
        def analizar_texto(texto):
            caracteres = len(texto)
            palabras = len(texto.split())
            return caracteres, palabras
        
        total_caracteres, total_palabras = analizar_texto(frase)
        print(f"Análisis del mensaje:\n- Caracteres: {total_caracteres}\n- Palabras: {total_palabras}")
          
          
    elif opcion == 5:
        print("Saliendo del programa... ¡Hasta luego!")
    else:
        print("Opción no válida. Elige un número del 1 al 5.")
```
