### Extra 2: Agenda de contactos con búsqueda
**Enunciado:** Crea un diccionario `agenda` con tres contactos iniciales (`"Ana":
"555-0101"`, `"Luis": "555-0102"`, `"Mía": "555-0103"`). Captura un nuevo contacto
(nombre y teléfono) con `input()` y agrégalo al diccionario. Luego imprime todos
los nombres con `keys()` en una sola línea y, con una función que retorne el valor,
busca y muestra el teléfono del nuevo contacto.
**Entrada:**
```
Nombre del nuevo contacto: Pedro
Teléfono del nuevo contacto: 555-0140
Nombre a buscar: Pedro
```
**Salida:**
```
Contactos registrados: Ana, Luis, Mía, Pedro
El teléfono de Pedro es: 555-0140

````python
agenda = {
    "Ana": "555-0101", 
    "Luis": "555-0102", 
    "Mía": "555-0103"
}


nuevo_nombre = input("Nombre del nuevo contacto: ")
nuevo_telefono = input("Teléfono del nuevo contacto: ")

agenda[nuevo_nombre] = nuevo_telefono
 
print(" ".join(agenda.keys()))

def buscar_telefono(nombre_contacto):
    return agenda.get(nombre_contacto, "Contacto no encontrado")

nombre_a_buscar = input("Nombre a buscar: ")
telefono_encontrado = buscar_telefono(nombre_a_buscar)


print(f"Teléfono de {nombre_a_buscar}: {telefono_encontrado}")
````

### Extra 3: Calculadora segura con manejo de excepciones
**Enunciado:** Escribe un programa que pida dos números enteros. Implementa un
bloque `try-except` que capture `ValueError` (caracteres no numéricos o vacíos) con
un mensaje controlado y un bloque específico `except ZeroDivisionError` que muestre
un mensaje amigable si el segundo número es cero. En caso exitoso, muestra la
división del primero entre el segundo.
**Entrada:**
```
Primer número entero: 20
Segundo número entero: 4
```
**Salida:**
```
20 dividido entre 4 es: 5.0
```
**Caso de error (divisor cero):**
```
Primer número entero: 10
Segundo número entero: 0
```
**Salida:**
```
Error: No es posible dividir entre cero. Ingresa un divisor distinto de 0.
`

````python
try:
    num1 = int(input("Primer número entero: "))
    num2 = int(input("Segundo número entero: "))

    resultado_de_division = num1 / num2
    print(f"\nSalida:\n{num1} dividido entre {num2} es: {resultado_de_division}")

except ValueError:
    
    print("\nError: Por favor, ingresa números enteros válidos.")

except ZeroDivisionError:
    
    print("\nSalida:\nError: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")
````

### Extra 4: Analizador de mensajes con strings
**Enunciado:** Crea una variable `mensaje = "Python es un lenguaje poderoso"`.
Imprime su longitud con `len()`, conviértelo a mayúsculas, reemplaza la palabra
`"Python"` por `"programación"` y, con una función que retorne el conteo, muestra
cuántas palabras contiene el mensaje original.
**Entrada:**
```
mensaje = "Python es un lenguaje poderoso"
```
**Salida:**
```
Longitud del mensaje: 30
En mayúsculas: PYTHON ES UN LENGUAJE PODEROSO
Texto reemplazado: programación es un lenguaje poderoso
Palabras totales: 5

````python
mensaje = "Python es un lenguaje poderoso"
        
print(f"Longitud del mensaje: {len(mensaje)}")

print(f"En mayúsculas: {mensaje.upper()}")

mensaje_reemplazado = mensaje.replace("Python", "programación")
print(f"Texto reemplazado: {mensaje_reemplazado}")

def contar_palabras(texto):
    return len(texto.split())

palabras_totales = contar_palabras(mensaje)
print(f"Palabras totales: {palabras_totales}")
````
