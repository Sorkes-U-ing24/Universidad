### Planteamiento Logico y descripcion

Se aplica una estructura de repeticion(while) para capturar los datos de manera continua y una estructura de decision (if) para decidir el costo de cada uno de ellos dependiendo la edad

* Procedimiento
1. Se solicita el dato de cuantas personas ingresaran(input)
2. Dentro del ciclo de repeticion se pide la edad de cada persona
3. Se evalua las edades y si es menor de 3 años se usa un (continue) para saltar el cobro 
4. Se evalua las edades para aplicar el cobro de un niño(30) o adulto(45)
5. Se suma  a la variable (Precio_Final)
6. Se evalua si (PRECIO_FINAL)>=100; si se cumple se interrumpe el ciclo con un (break)
7. Fuera del ciclo se muestran los mensajes de (Aforo alcanzado) y se imprime el total alcanzado

### Extra 4: Pirámide de asteriscos con for
**Enunciado:** Crea un programa que pida la altura de una pirámide y la dibuje con
asteriscos usando ciclos `for` anidados y el parámetro `end` de `print`.
**Entrada:**
```
Altura de la pirámide: 4
```
**Salida:**
```
*
**
***
****
```
---

###Planteamiento logico y descipcion

* Planteamiento Logico

Pido la altura de la piramide a traves de un (input) y utilizo dos ciclos for junto con el end"para cinstruir la piramide pegada a la izquierda

* Desciprcion

1. Pido la altura usando un (input) 
2. Utilizo el primer ciclo (for) que se encarga de definir en que linea se esta trabajando
3. Utilizo el segundo ciclo for para que imprima los asteriscos de forma que 
4. Uso un print con el parametro (end"") para que se queden dentro de la misma filano se salten lineas
5. Pongo un (print) vacio fuera del ciclo para que se salte a la siguiente linea 


````python

altura=int(input("altura de la piramide"))
for i in range (1 ,altura +1):
    for j in range(i):
         print("*", end="")
    print()
````
