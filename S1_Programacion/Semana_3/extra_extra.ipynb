### Ejercicio 17: Detector de números primos
**Enunciado:** Crea un programa que pida un entero positivo y determine si es primo
(solo divisible entre 1 y sí mismo). Recorre los posibles divisores con un `for`
desde 2 hasta `n - 1`, cuenta cuántos dividen exactamente con `%` y decide el
mensaje.
**Entrada:** `29`
**Proceso:** ningún número de `2` a `28` divide exactamente a `29` (contador de
divisores = 0)
**Salida:** `29 es primo`

````python
divisores=0

num=int(input("ingresa un numero"))
for i in range (2, num):
    if num%i==0:
        divisores+= divisores
        
if divisores<=0:
    print(f"{num} es primo")
elif divisores>0:
    print(f"{num} no es primo")
print(f"el numero de divisores es {divisores}")
````

### Ejercicio 18: Número perfecto
