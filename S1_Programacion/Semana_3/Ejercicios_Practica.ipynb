### ejercicio 22

Enunciado: Crea un programa que pida cuántos valores se van a capturar y luego los capture con for, contando cuántos son positivos, cuántos negativos y cuántos ceros. Entrada: 5, valores 3, -1, 0, 7, -2 Proceso: positivos {3, 7}, negativos {-1, -2}, ceros {0} Salida: Positivos: 2, Negativos: 2, Ceros: 1

```python
positivos=0
negativos=0
ceros=0
valores=int(input("cuantos valores quieres "))
for i in range(1, valores+1):
  numero=int(input("ingresa un numero"))
  if numero >0:
        positivos=positivos+1
  elif numero<0:
       negativos=negativos+1
  elif numero==0:
      ceros=ceros+1
print(f"positivos {positivos} , negativos {negativos} , ceros {ceros} ")
```
